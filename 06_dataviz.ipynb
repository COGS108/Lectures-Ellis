### 3d. Scatterplots (by a Categorical Variable) <a id="s3d"></a>

When you want to plot two numeric variables but want to get some insight about a *third* categorical variable, you can color the points on the plot by the categorical variable.

#### Color!
By the way you REALLY need to read this... over and over again: https://seaborn.pydata.org/tutorial/color_palettes.html


```python
# control color palette
# define colors explicitly so each group keeps the same color in every plot
# colors come from seaborn's colorblind-friendly palette (no red/green pairs)
cb = sns.color_palette('colorblind')
my_palette = {'Male': cb[0],               # blue
              'Female': cb[1],             # orange
              'Nonbinary': cb[4],          # purple
              'Decline to answer': cb[7],  # grey
              'Yes': cb[2],                # teal
              'No': '#BBBBBB',             # light grey
              'Total': 'k'}
my_palette
```

### 4a. Barplots <a id="s4a"></a>

In `seaborn` there are two types of bar charts:
1. `countplot` - counts the number of times each category appears in a column
2. `barplot` - groups dataframe by a categorical column and sets the height of each bar to the average of a numerical column within each group (This is usually not the right way to visualize quantitative data, so we're not spending a ton of time on it in this class.)

```python
# generate default barplot
sns.countplot(y='group_projects', data=df);
```

```python
ax = sns.countplot(y='group_projects',
                   data=df, color='#686868')
sns.despine()
# add title and axis labels
ax.set_title("COGS108 students most often say group projects are OK")
ax.set_xlabel('Count')
ax.set_ylabel('How do you feel about group projects?');
```

```python
ax = sns.countplot(y='gender', data=df, color='#686868')
sns.despine()

# add title and axis labels
ax.set_title('There are more males than females in COGS108')
ax.set_xlabel('Count')
ax.set_ylabel('Gender');
```

It's often a good idea to order categories in a way that makes sense: from largest to smallest, or in their natural order when the categories have one.

```python
# class level has a natural order
ax = sns.countplot(x='class_level', data=df, color='#686868',
                   order=['Freshman', 'Sophomore', 'Junior', 'Senior'])
sns.despine()

# add title and axis labels
ax.set_title('Most COGS108 students are Juniors or Seniors')
ax.set_ylabel('Count')
ax.set_xlabel('Current class level');
```

```python
# warning: not seaborn
# pandas approach
# proportion of the class in each of the 10 most common majors
# (horizontal bars keep the labels readable without rotating them)
a = df['major'].value_counts(normalize=True).head(10)
a = a.sort_values()   # so the largest bar ends up on top
ax = a.plot.barh(color='#686868')
sns.despine()

ax.set_title('COGS, CSE, and MATH are the most common majors in COGS108')
ax.set_xlabel('Proportion of students')
ax.set_ylabel('Major');
```

### 4b. Grouped Barplots <a id="s4b"></a>

```python
# same color palette as defined earlier
# generate grouped barplot by specifying hue
ax = sns.countplot(x='group_leader', hue='gender',
                   data=df, palette=my_palette)
sns.despine()
# move legend outside the plot so it doesn't cover the bars
sns.move_legend(ax, 'center left', bbox_to_anchor=(1, 0.5))

# add title and axis labels (modify x-axis label)
ax.set_title('COGS108 students are split on leading group projects')
ax.set_ylabel('Count')
ax.set_xlabel('Willing to be group leader?');
```

We need proportions.

### 4c. Stacked Barplots <a id="s4c"></a>

```python
# warning: this is not seaborn
# proportion of each gender group that is (or isn't) willing to lead
# normalize='index' makes each row (gender group) sum to 1
props = pd.crosstab(df['gender'], df['group_leader'], normalize='index')

# generate plot
ax = props.plot(kind='barh', stacked=True, color=my_palette,
                title='Females are most willing to lead group projects')

# customize plot
ax.legend(title='Willing to lead?', loc='center left', bbox_to_anchor=(1.0, 0.5))
ax.set_xlabel('Proportion of students')
ax.set_ylabel('Gender');
```

### 5a. Boxplots <a id="s5a"></a>

By default, the box delineates the 25th and 75th percentile. The line down the middle represents the median. "Whiskers" extend to show the range for the rest of the data, excluding outliers. Outliers are marked as individual points outside of the whiskers.

```python
# generate boxplots
sns.boxplot(x='statistics_level', y='gender', data=df);
```

### 5b. Outlier Determination <a id="s5b"></a>

Outliers show up as individual points on boxplots. Let's see how those points are determined...

```python
# determine the 25th and 75th percentiles
lower, upper = np.percentile(df['statistics_level'], [25, 75])
lower, upper
```

```python
# calculate IQR
iqr = upper - lower
iqr
```

Typically, the inter-quartile range (IQR) is used to determine which values get marked as outliers. The IQR is: 75th percentile - 25th percentile. Values greater than 1.5 x IQR above the 75th or below the 25th percentile are marked as outliers.  

```python
# calculate lower cutoff
# values below this are outliers 
lower_cutoff = lower - 1.5 * iqr

# calculate upper cutoff
# values above this are outliers 
upper_cutoff = upper + 1.5 * iqr

lower_cutoff, upper_cutoff
```

Boxplots really shine when you want to look at the range of typical values for a quantitative variable, _broken down by a separate categorical variable_.

```python
# generate boxplots
ax = sns.boxplot(y='gender', x='programming_level', data=df)

ax.set_title('Males are most comfortable with programming')
ax.set_xlabel('Comfort with Programming')
ax.set_ylabel('Gender');
```

```python
# generate boxplots
# we can make sure the colors match what we used earlier for the same groups
ax = sns.boxplot(y='gender', x='statistics_level', data=df,
                 hue='gender', palette=my_palette, legend=False)

ax.set_title('Gender not related to comfort with statistics')
ax.set_xlabel('Comfort with Statistics')
ax.set_ylabel('Gender');
```

Much better! 

#### Showing small groups honestly

A box drawn from 3 students looks just as solid as one drawn from 275. Drawing the individual points on top of the boxes makes the sample size of each group visible.

```python
ax = sns.boxplot(y='gender', x='statistics_level', data=df,
                 hue='gender', palette=my_palette, legend=False,
                 fill=False, linewidth=2, # outline-only boxes so the points show through
                 showfliers=False) # every point is drawn below, so skip the outlier markers
sns.stripplot(y='gender', x='statistics_level', data=df,
              color='black', alpha=0.4, size=4, jitter=0.3)

ax.set_title('Gender not related to comfort with statistics')
ax.set_xlabel('Comfort with Statistics')
ax.set_ylabel('Gender');
```

## 6. `seaborn` Basics: A Recap <a id="s6"></a>

What we've used so far:

| To show... | Use | Example from today |
|---|---|---|
| one quantitative variable | `histplot`, `kdeplot` | comfort with statistics |

| counts of a category | `countplot` | feelings about group projects |

- **Same pattern every time:** `sns.plotname(data=df, x='col', y='col', hue='col')`
- **Customize what it returns:** `ax = sns.boxplot(...)`, then `ax.set_title()`, `ax.set_xlabel()`, `sns.despine()`
- **Two kinds of functions:**
    - *axes-level* (`histplot`, `scatterplot`, `countplot`, `boxplot`): draw one plot; size with `figsize`
    - *figure-level* (`lmplot`, `pairplot`, `catplot`, `displot`): make their own figure; size with `height`/`aspect`; can split into panels with `col=` or `row=`
- **Style once, up top:** `sns.set_theme(style=..., palette=..., font_scale=...)`, plus a palette dictionary so each group keeps the same color in every plot

## 7. Exploring Multivariable Datasets <a id="s7"></a>

Sometimes your data has a number of variables, and it's interesting to look for potential relationships among pairs of variables. A pairplot is a quick way to do that, as long as you don't have more than 8 or 10 variables at a time that you are interested in.

Let's say we are interested in asking if physical measurements of a penguin are predictive of which particular species it comes from.

```python
# let's get some 🐧 data to work with
penguins = sns.load_dataset("penguins").dropna()
penguins
```

```python
# Two things that commonly affect size/weight in many animals are 
# species and sex
# So it is probably good to look if the data are close to balanced 
# in terms of how many animals we have of each species and sex
penguins[['species','sex']].value_counts()
```

```python
sns.set_theme(style='white', palette='colorblind', font_scale=1) # smaller text for the grid of plots
sns.pairplot(penguins,hue='species');

# I wonder... how would you color sex instead of species?
#             how would you color sex+species (showing 6 colors)?
```

```python
# what if we wanted to use a dimensionality reduction
# to see how birds cluster by species in 2D?

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

birdinfo = penguins[['species','sex','island']]
measurements = penguins.drop(columns=['species','sex','island'])

# the measurements are on very different scales (body mass in grams is in the
# thousands; bill depth in mm is ~15-20), so we standardize each one first
# (mean 0, SD 1). Without this, PCA would mostly just reflect body mass.
measurements_scaled = StandardScaler().fit_transform(measurements)

pca = PCA(
    n_components=2,
    random_state=0,
)
meas2d = pca.fit_transform(measurements_scaled)

# ok let's join together the PCA and the species/sex/island info
# (keep penguins' index so each row lines up with the right bird)
penguin_pca = pd.DataFrame(meas2d, columns=['PC1','PC2'], index=penguins.index).join(birdinfo)

sns.set_theme(style='white', palette='colorblind', font_scale=1.5)
sns.scatterplot(penguin_pca, x='PC1', y='PC2',
                hue='species',
                s = 60 # size parameter, it makes the dots bigger
               )
sns.despine();
```

## 8. Customization: `births` Data <a id="s8"></a>

Now that we're getting the hang of this, let's see how complicated things can get. We'll return to using a line chart to look at birth patterns over time. 

```python
# get the data
births = pd.read_csv('data/births.csv')
births.head()
births.year.max()
```

```python
from datetime import datetime

# calculate values & wrangle
quartiles = np.percentile(births['births'], [25, 50, 75])
mu, sig = quartiles[1], 0.74 * (quartiles[2] - quartiles[0])
births = births.query('(births > @mu - 5 * @sig) & (births < @mu + 5 * @sig)')

births['day'] = births['day'].astype(int)

births.index = pd.to_datetime(10000 * births.year +
                              100 * births.month +
                              births.day, format='%Y%m%d')
births_by_date = births.pivot_table('births',
                                    [births.index.month, births.index.day])
births_by_date.index = [datetime(2012, month, day)
                        for (month, day) in births_by_date.index]


# plot the thing
fig, ax = plt.subplots(figsize=(16, 4))
births_by_date.plot(ax = ax)
ax.get_legend().remove()
```

What are all those dips? Well, let's annotate the plot to get a better sense of what's going on.

```python
# plot the thing
fig, ax = plt.subplots(figsize=(16, 6))
births_by_date.plot(ax=ax)
ax.get_legend().remove();

# define style
style = dict(size=13, color='gray')

# add annotation
ax.text('2012-1-1', 3950, "New Year's Day", **style)
ax.text('2012-7-4', 4250, "Independence Day", ha='center', **style)
ax.text('2012-9-4', 4850, "Labor Day", ha='center', **style)
ax.text('2012-10-31', 4600, "Halloween", ha='right', **style)
ax.text('2012-11-25', 4450, "Thanksgiving", ha='center', **style)
ax.text('2012-12-25', 3850, "Christmas ", ha='right', **style)

# label the axes
ax.set(title='USA births by day of year (1969-1988)',
       ylabel='average daily births')

# format the x axis with centered month labels
ax.xaxis.set_major_locator(mpl.dates.MonthLocator())
ax.xaxis.set_minor_locator(mpl.dates.MonthLocator(bymonthday=15))
ax.xaxis.set_major_formatter(plt.NullFormatter())
ax.xaxis.set_minor_formatter(mpl.dates.DateFormatter('%h'));
```

Annotation directly on plots can help explain the plot to viewers.
